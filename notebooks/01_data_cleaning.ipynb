```python
import pandas as pd 
from pathlib import Path
```

```python
# Carga de datos
ruta_actual = Path().resolve()
ruta_proyecto = ruta_actual.parent

RAW_PATH = ruta_proyecto / "data" / "raw" / "Lima_properties_raw.csv" 
PROCESSED_PATH = ruta_proyecto / "data" / "processed" / "lima_properties_clean.csv"

PROCESSED_PATH.parent.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(RAW_PATH)

print(f"Los registros originales son {len(df)}")

df.head(10)
```

```python
#Tratamiento de datos

#Eliminacion de duplicados
df = df.drop_duplicates(subset = ["listing_id"])

df.columns
```

```python
# Visualizando el tipo de dato
print(df.dtypes)
print("\n")
print(df.isnull().sum())
```

```python
#Eliminando columnas que no aportan señal predictiva
columnas_no_predictivas = ['listing_url', 'scraping_date', 'source', 'price_usd', 'json_name']
df = df.drop(columnas_no_predictivas, axis =1)

print(f"Las columnas restantes son {df.shape[1]}")
print(df.columns.to_list())
```

```python
#Eliminando columnas
df = df.drop("district", axis = 1)

#Renombrando columnas
dic = {
    "district_scraped" : "district"
}
df = df.rename(dic, axis = 1)

df
```

```python
#Filtrando outliers area total

print(df["area_total"].describe())

print(f"\nValores fuera de rango (< 20 o >1000): {((df['area_total']<20) | (df['area_total']>1000)).sum()}")
```

```python
mask = ~((df["area_total"]<20)|(df["area_total"]>1000))
df = df[mask]

print(f"La cantidad de filas es {len(df)}")
```

```python
#Tratammiento de NaNs parking_spaces y maintenance_fee (ausencia de valor quiere decir que no tienen esa propiedad)

df["maintenance_fee"] = df["maintenance_fee"].fillna(0)
df["parking_spaces"] = df["parking_spaces"].fillna(0)

df.isna().sum()
```

```python
#Nuestro target es el precio, en el target nunca se imputa, por lo que eliminaresmos los nan
df= df.dropna(subset = ["price_pen"])
len(df)
```

```python
print(df.groupby("district")["area_total"].count().sort_values())
#Dataset desbalanceado, para futuro recolectar datos de otros portales
#Tomar en consideracion que probablemente haya departamentos duplicados por lo que no se podrá 
```

```python
df["area_total"] = df["area_total"].fillna(
    df.groupby("district")["area_total"].transform("median")
)

df["bedrooms"] = df["bedrooms"].fillna(
    df.groupby("district")["bedrooms"].transform("median")
)

df["bathrooms"] = df["bathrooms"].fillna(
    df.groupby("district")["bathrooms"].transform("median")
)
df.isnull().sum()
```

```python
# Convirtiendo floats a entero
df["bathrooms"] = df["bathrooms"].astype(int)
df["bedrooms"] = df["bedrooms"].astype(int)
df["parking_spaces"] = df["parking_spaces"].astype(int)

df.info()
```

```python
df.to_csv(PROCESSED_PATH,index=False, encoding="utf-8-sig" )
from pathlib import Path
print(f"Archivo guardado en: {PROCESSED_PATH}")
print(f"Existe: {PROCESSED_PATH.exists()}")
print(f"Tamaño: {PROCESSED_PATH.stat().st_size / 1024:.1f} KB")
```
